```python
file_name = 'housepriceraw.csv'
processed_date = '2026-02-24'
```

```python
adfs_path='abfss://house_price_DEV@onelake.dfs.fabric.microsoft.com/houseprice_LH.Lakehouse/Files/raw'
```

```python
full_path= f"{adfs_path}/{file_name}"
print(full_path)
```

```python
df = spark.read.csv(path= full_path,header=True,inferSchema=True)
display(df)
```

```python
from pyspark.sql.functions import lit
if df.count()>1:
    print("This file has data")
    df_new = df.withColumn("Processing_date",lit(processed_date))
    # display(df_new)
    df_new.write.format('csv').option("header","True").mode("append").partitionBy("Processing_date").save("abfss://house_price_DEV@onelake.dfs.fabric.microsoft.com/houseprice_LH.Lakehouse/Files/Landing")
    print("Data Loading to landing zone successfully")
else:
    print("File contain no Data")
```
